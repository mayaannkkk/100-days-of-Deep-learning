```python
import numpy as np

docs = ['go india',
        'india india',
        'hip hip hurray',
        'jeetega bhai jeetega india',
        'bharat mata ki jai',
        'kohli kohli',
        'sachin sachin',
        'dhoni dhoni',
        'modi ji ki jai',
        'inquilab zindabad']
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index
```

```python
sequence = tokenizer.texts_to_sequences(docs)
sequence
```

```python
len(tokenizer.word_index)
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequence,padding='post')
```

```python
sequences
```

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import *
model = Sequential([
    Embedding(18,output_dim=2)
])
```

```python
model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
pred
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

model1 = Sequential([
    Embedding(
        input_dim=18,
        output_dim=2,
        input_shape=(4,)
    ),
    SimpleRNN(32),
    Dense(1, activation='sigmoid')
])

model1.summary()
```

```python
model1.compile('adam', 'mse')
```

```python
embedding_weights = model1.layers[0].get_weights()[0]
```

```python
print("\nEmbedding Matrix Shape:")
print(embedding_weights.shape)

print("\nEmbedding Matrix:\n")
print(embedding_weights)
```
